## Task 5 — Conditional Probability, Revisited


```python
p_default = (df["defaulted"] == "Yes").mean()

print("P(Default):", p_default)

rate_by_collateral = df.groupby("has_collateral")["defaulted"].apply(
    lambda x: (x == "Yes").mean()
)

print(rate_by_collateral)
```

1. Report P(Default), P(Default | Has Collateral) and P(Default | No Collateral)

- P(Default) = 0.19344 or 19.34%
- P(Default | Has Collateral) = 0.120805 or 12.08%.
- P(Default | No Collateral) = 0.260000 or 26.00%.
